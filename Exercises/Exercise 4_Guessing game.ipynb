```python
import time
name = "Aku"
i = 0

print("Guess my name")
time.sleep(0.5)
guess = input()

if guess == name:
    print("Congratulation!")
    print("First try!")
while guess != name:
    print("Incorrect")
    print("Input n to quit")
    i += 1
    guess = input()

    if guess == "n":
        break
    if guess == name:
        i += 1
        print("Congratulation!")
        print("Guesses:", i)
```
