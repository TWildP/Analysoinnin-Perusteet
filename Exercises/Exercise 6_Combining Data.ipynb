```python
import pandas as pd

data1 = {"order_id": [2,1,3], "item_name": ["Burger","Pizza","Salad"], "quantity": [2,1,3]}
orders_df = pd.DataFrame(data1)

data2 = {"order_id": [4,6,5], "item_name": ["Soup","Burger","Wok"], "quantity": [1,3,4]}
portions_df = pd.DataFrame(data2)

pd.concat([orders_df.sort_values("order_id"),portions_df.sort_values("order_id")])
# Järjestin datan order_id:n mukaan
```

```python
import pandas as pd

data1 = {"order_id": [2,1,3],
         "customer_name": ["Alice","Bob","Charlie"],
         "total_cost": [25.5,30.2,15.75]}
orders_df = pd.DataFrame(data1)

data2 = {"order_id": [2,1,3],
         "item_name": ["Burger","Pizza","Salad"],
         "quantity": [2,1,3]}
portions_df = pd.DataFrame(data2)

#pd.merge(orders_df,portions_df)
pd.merge(orders_df.sort_values("order_id"),portions_df)
```
